```python
history = pd.read_csv("../models/artifacts/ppo/depth_1_5_onehot/v001/history.csv")

history_5 = history[history['curriculum_depth'] == 1]

print(history_5.shape)
solve_rate = history.copy()

solve_rate['avg_solve'] = solve_rate.groupby('curriculum_depth')['train_solve_rate'].transform('mean')

display(solve_rate[['curriculum_depth', 'avg_solve']])

depth_stats = solve_rate.groupby('curriculum_depth').agg(
    depth_count=('curriculum_depth', 'count'),
    avg_solve=('train_solve_rate', 'mean')
)

display(depth_stats)

display(history_5.head())
print(history.shape)
print(history['curriculum_depth'].value_counts())

fig1 = px.line(history_5, x='timesteps', y='train_solve_rate')
fig1.show()
```
